## Section 1: Understanding Agent File Structure

### Copilot Agent File Requirements
Copilot agents are defined in markdown files with YAML frontmatter followed by agent instructions in the body.

**File Location**: `.github/agents/*.agent.md`  
**Naming Convention**: `{agent-name}.agent.md` (e.g., `labeling.agent.md`)

### Required Frontmatter Properties (Copilot Spec)
```yaml
---
name: unique-agent-name                  # Optional, defaults to filename
description: "Clear description"          # REQUIRED - agent purpose and capabilities
target: vscode                            # Optional: vscode, github-copilot, or both
tools: ['read', 'edit', 'search']        # Optional: tool aliases the agent can use
mcp-servers:                              # Optional: MCP server config (org/enterprise only)
  server-name:
    type: local
    command: command-to-run
metadata:                                 # Optional: custom key-value pairs
  key: value
---
```

### LightSpeed Extended Frontmatter (EXISTING FIELDS TO PRESERVE)
```yaml
---
file_type: agent
name: agent-name
title: "Agent Title"
description: "Description"
version: "v1.0"
last_updated: "YYYY-MM-DD"
owners: ["lightspeedwp/team"]
tags: ["tag1", "tag2"]
category: automation/documentation/quality/etc
status: active/deprecated/planning
visibility: public/internal
references:
  - path: scripts/agent-file.js
    description: "Script implementation"
  - path: schemas/agent.schema.json
    description: "Schema definition"
---
```

### Relationship Map
```
Agent File (.agent.md)
├── Frontmatter (YAML)
│   ├── description → Agent's purpose
│   ├── tools → Capabilities
│   ├── references → Links to implementation
│   └── handoffs → Orchestration config
│
├── Implementation (JavaScript)
│   └── .github/agents/{agent-name}.agent.js
│       ├── Imports from .github/agents/includes/
│       └── Exports main run() function
│
├── Workflow Triggers
│   └── .github/workflows/{workflow-name}.yml
│       └── Uses agent via JavaScript
│
├── Schema Definitions
│   └── schemas/{schema-name}.json
│       └── Validates agent configuration
│
└── Body Content
    ├── Guardrails section
    ├── Process documentation
    └── Example usage
```

## Section 3: Supported Tool Aliases and Configuration

### Copilot Tool Aliases (Reference)
Agents can declare which of these tools they can use:

| Alias | Compatible Names | Purpose | Use Case |
|-------|------------------|---------|----------|
| `shell` | Bash, powershell | Execute shell commands | Automation, builds |
| `read` | Read, NotebookRead | Read file contents | Code analysis, review |
| `edit` | Edit, MultiEdit, Write, NotebookEdit | Modify files | Implementation, fixes |
| `search` | Grep, Glob | Search files/text | Finding references, patterns |
| `custom-agent` | Task | Invoke other agents | Orchestration, delegation |
| `web` | WebSearch, WebFetch | URL fetching, web search | Research, external data |
| `todo` | TodoWrite | Manage task lists | Task tracking, planning |
| `github/*` | GitHub MCP tools | Repository operations | PR/issue management |
| `playwright/*` | Playwright testing | E2E test automation | Testing workflows |

### MCP Server Configuration Example
```yaml
mcp-servers:
  github:
    tools: ["*"]  # Enable all GitHub tools
  playwright:
    tools: ["playwright/navigate", "playwright/fill"]  # Specific tools
```

### Agent Body Structure (After Frontmatter)

```markdown
## Purpose & Capabilities

Brief description of what the agent does, why it exists, and key capabilities.

## Workflow Integration

- **Trigger**: `.github/workflows/agent-name.yml`
- **Event**: Push to develop branch, pull_request events
- **Permissions**: contents:read, issues:write, pull-requests:write

## Implementation

### Main Entry Point
- **File**: `.github/agents/agent-name.agent.js`
- **Function**: `runAgent(context, options)`

### Shared Utilities
- **Location**: `.github/agents/includes/`
- **Files Used**: List relevant include files
- **Dependencies**: List external dependencies

## Tools Configuration

### Required Tools
- **`read`**: Examine file contents and PR/issue details
- **`edit`**: Modify files and add/update comments
- **`search`**: Find related code and patterns
- **`github/*`**: Access GitHub API for repository operations

### Tool Limitations
- `read`: Limited to repository files, no external access
- `edit`: Cannot modify protected files
- `search`: Searches only within repository
- `github/*`: Scoped to source repository token

## Guardrails

**Critical Constraints**:
1. Never output secrets or sensitive data
2. Always validate input before processing
3. Implement error handling with graceful degradation
4. Log all operations for audit trail
5. Support dry-run mode for testing

**Limitations**:
- Cannot access external APIs without explicit MCP server
- Cannot execute arbitrary shell commands
- Cannot modify GitHub workflow files
- Cannot delete resources (only mark as deprecated)

**Safety Measures**:
- All operations are reversible where possible
- Validation gates before any state changes
- Automated rollback on critical errors
- Rate limiting to prevent API abuse

## Process

1. **Analysis Phase**: Examine context and requirements
2. **Validation Phase**: Verify all preconditions met
3. **Action Phase**: Execute intended operations
4. **Verification Phase**: Confirm success and log results
5. **Notification Phase**: Communicate outcomes

## Error Handling

| Error Type | Recovery Strategy |
|-----------|------------------|
| API Rate Limit | Retry with exponential backoff |
| Missing Permissions | Log error and skip operation |
| Invalid Input | Return early with error message |
| Network Timeout | Retry up to 3 times |
| Dry-run Enabled | Log planned actions without executing |

## Testing & Validation

- **Unit Tests**: `.github/agents/__tests__/{agent-name}.test.js`
- **Local Testing**: Can be tested in VS Code locally
- **Validation**: Use `node .github/agents/{agent-name}.agent.js --dry-run`

## References

- [GitHub Copilot Agent Documentation](https://code.visualstudio.com/docs/copilot/customization/custom-agents)
- [LightSpeed Agent Guidelines](../../AGENTS.md)
- [Frontmatter Schema](../../schemas/frontmatter.schema.json)
- [Automation Governance](../../.github/AUTOMATION_GOVERNANCE.md)
```
